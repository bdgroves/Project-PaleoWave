```python
## 3. Cluster Analysis — Are These Truly Distinct Localities?
# LIT-001 through 005 are all Augusta Mtns / Favret Canyon.
# We need to decide: treat as one training point or multiple?

new_locs = df_lit[~df_lit.in_pbdb].copy()

# Pairwise distances within new localities
print('=== Pairwise distances (km) between new localities ===')
for i, row_i in new_locs.iterrows():
    for j, row_j in new_locs.iterrows():
        if j <= i: continue
        dist = math.sqrt(
            (row_i.latitude  - row_j.latitude)**2  * 111**2 +
            (row_i.longitude - row_j.longitude)**2 * (math.cos(math.radians(row_i.latitude))*111)**2
        )
        print(f'  {row_i.locality_id} ↔ {row_j.locality_id}: {dist:.1f} km   ({row_i.taxon[:20]} / {row_j.taxon[:20]})')

print()
# Compare to existing PBDB cluster
pbdb = pd.read_csv(PBDB_DIR / 'pbdb_occurrences_clean.csv')
print('=== Distance from Augusta Mtns centroid to existing PBDB points ===')
aug_lat, aug_lon = 40.0, -117.54
pbdb['dist_aug'] = pbdb.apply(lambda r: math.sqrt(
    (r.latitude - aug_lat)**2 * 111**2 +
    (r.longitude - aug_lon)**2 * (math.cos(math.radians(aug_lat))*111)**2
), axis=1)
print(pbdb.sort_values('dist_aug')[['occurrence_id','taxon_name','formation','latitude','longitude','dist_aug']].head(8).to_string(index=False))
```

```python
## 4. Deduplicate Against PBDB (5km radius)
# Any literature locality within 5km of an existing PBDB point is a duplicate.

DEDUP_RADIUS_KM = 5.0

def min_dist_to_pbdb(lat, lon, pbdb_df):
    dists = pbdb_df.apply(lambda r: math.sqrt(
        (r.latitude - lat)**2 * 111**2 +
        (r.longitude - lon)**2 * (math.cos(math.radians(lat))*111)**2
    ), axis=1)
    idx = dists.idxmin()
    return dists[idx], pbdb_df.loc[idx, 'occurrence_id'], pbdb_df.loc[idx, 'formation']

new_locs = df_lit[~df_lit.in_pbdb].copy()
results = new_locs.apply(
    lambda r: pd.Series(min_dist_to_pbdb(r.latitude, r.longitude, pbdb),
                        index=['min_dist_pbdb_km', 'nearest_pbdb_id', 'nearest_pbdb_formation']),
    axis=1)
new_locs = pd.concat([new_locs, results], axis=1)

usable = new_locs[new_locs.min_dist_pbdb_km >= DEDUP_RADIUS_KM].copy()
dup    = new_locs[new_locs.min_dist_pbdb_km <  DEDUP_RADIUS_KM].copy()

print(f'New literature localities:    {len(new_locs)}')
print(f'  Within 5km of PBDB (skip):  {len(dup)}')
print(f'  >5km from PBDB (usable):    {len(usable)}')
print()
print('=== DUPLICATED BY PBDB ===')
print(dup[['locality_id','taxon','latitude','longitude','min_dist_pbdb_km','nearest_pbdb_id','nearest_pbdb_formation']].to_string(index=False))
print()
if len(usable):
    print('=== GENUINELY NEW ===')
    print(usable[['locality_id','taxon','formation','latitude','longitude','uncertainty_m','min_dist_pbdb_km']].to_string(index=False))
else:
    print('No genuinely new localities — all covered by PBDB within 5km.')
    print('PBDB coverage is more complete than expected for the Augusta Mountains.')
```

```python
## 5. Data Quality Audit — PBDB Formation Field Gaps
# The literature harvest found no new localities, but it revealed PBDB quality issues.
# Cross-reference literature source formations against PBDB formation fields.

print('=== Literature vs PBDB Formation Cross-Check ===')
print()
for _, row in new_locs.iterrows():
    pbdb_form = str(row.nearest_pbdb_formation).strip()
    lit_form  = row.formation
    match = '✓' if pbdb_form.lower() == lit_form.lower() else ('NaN→FIX' if pbdb_form == 'nan' else '≠ CHECK')
    print(f'  {row.locality_id}  {row.nearest_pbdb_id:15s}  '
          f'PBDB={pbdb_form:12s}  LIT={lit_form:8s}  {match}')

print()
# Find all PBDB points with missing formation that the literature can fill
fix_needed = new_locs[
    new_locs.nearest_pbdb_formation.isna() |
    (new_locs.nearest_pbdb_formation.astype(str).str.strip() == 'nan')
].copy()

if len(fix_needed):
    print(f'=== PBDB RECORDS MISSING FORMATION ({len(fix_needed)} found) ===')
    print('These can be corrected in PBDB using literature sources:')
    for _, row in fix_needed.iterrows():
        print(f'  {row.nearest_pbdb_id}: should be formation="{row.formation}"')
        print(f'    Source: {row.source}')
        print(f'    Basis:  {row.basis}')
        print()
else:
    print('No missing formation fields detected.')
```

```python
## 6. Apply Formation Fix and Save

# Save the full literature table for reference
df_lit.to_csv(PBDB_DIR / 'literature_localities.csv', index=False)
print('Saved: literature_localities.csv')

# Apply formation fix to paleowave_combined.csv
combined = pd.read_csv(PBDB_DIR / 'paleowave_combined.csv')
print(f'Combined dataset before fix: {len(combined)} records')
print(f'  Missing formation: {combined.formation.isna().sum()}')

# Fix: occ:1186493 C. nevadanus at New Pass → formation = Prida
# (and any other NaN-formation records that the literature can confirm)
fixes = {
    'occ:1186493': 'Prida',   # C. nevadanus, New Pass, Merriam 1908
}
for occ_id, form in fixes.items():
    mask = combined.occurrence_id == occ_id
    if mask.sum() > 0:
        combined.loc[mask, 'formation'] = form
        print(f'  Fixed: {occ_id} → formation = {form}')
    else:
        print(f'  Not found in combined: {occ_id} (may be PBDB-only, fix in pbdb_occurrences_clean.csv)')

# Also fix pbdb_occurrences_clean.csv
pbdb_clean = pd.read_csv(PBDB_DIR / 'pbdb_occurrences_clean.csv')
for occ_id, form in fixes.items():
    mask = pbdb_clean.occurrence_id == occ_id
    if mask.sum() > 0:
        pbdb_clean.loc[mask, 'formation'] = form
        print(f'  Fixed in pbdb_clean: {occ_id} → formation = {form}')

combined.to_csv(PBDB_DIR / 'paleowave_combined.csv', index=False)
pbdb_clean.to_csv(PBDB_DIR / 'pbdb_occurrences_clean.csv', index=False)
print()
print(f'Combined dataset after fix: {len(combined)} records')
print(f'  Missing formation: {combined.formation.isna().sum()}')
print()
print('Saved: paleowave_combined.csv (formation fixed)')
print('Saved: pbdb_occurrences_clean.csv (formation fixed)')
```
